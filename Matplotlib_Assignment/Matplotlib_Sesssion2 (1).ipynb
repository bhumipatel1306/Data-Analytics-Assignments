```python
import matplotlib.pyplot as plt
import pandas as pd
```

1.Use plt.plot() to create a line chart showing the number of daily steps you walked over the last 7 days. Set the line color to green and linewidth to 2.

```python
days= ["Monday", "Tuesday", "Wednesday", "Thursday","Friday", "Saturday", "Sunday"]
steps= [3000, 4500, 5000, 4000, 6500, 8000, 7000]

plt.plot(days,steps,color="green",linewidth=2,marker="o")
plt.xlabel("Days")
plt.ylabel("Steps")
plt.title("My Steps Per Day")
plt.show()
```

2.Plot a line chart of your favorite music app's monthly active users for 6 months using plt.plot(). Add circle markers at each data point and set the marker color to red.<br><br><em><strong>Hint:</strong> Use the marker and markerfacecolor parameters in plt.plot().</em>

```python
months = ["January", "February", "March","April", "May", "June"]
users = [50, 55, 60, 58, 70, 80]

plt.plot(months,users,color="red",marker="o",markerfacecolor="red")
plt.xlabel("Months")
plt.ylabel("Users")
plt.title("My Music App Users")
plt.show()
```

3.Create a line chart comparing the number of orders you placed on Zomato and Swiggy over the last 5 months. Use different line styles and colors for each app, and add a legend to distinguish them.

```python
months = ["January", "February", "March", "April", "May"]
zomato = [8, 10, 6, 12, 15]
swiggy = [5, 7, 9, 8, 11]

plt.plot(months,zomato,color="red",linestyle="--",label="Zomato")
plt.plot(months,swiggy,color="green",linestyle="--",label="Swiggy")
plt.xlabel("Months")
plt.ylabel("Orders")
plt.title("My Orders Per Month",color="r",fontsize=15,pad=10)
plt.grid()
plt.legend()
plt.show()
```

4.Add a main title 'Monthly Data Usage Trend' and label the X-axis as 'Month' and Y-axis as 'Data Used (GB)' to a line chart showing your mobile data usage for the past 6 months.<br><br><em><strong>Hint:</strong> Use plt.title(), plt.xlabel(), and plt.ylabel().</em>

```python
months = ["June", "July", "August","September", "October", "November"]
data_used = [15, 20, 18, 25, 30, 28]

plt.plot(months,data_used,color="green",linestyle="--")
plt.xlabel("Months",color="r",fontsize=13)
plt.ylabel("Data Used (GB)",color="r",fontsize=13)
plt.title("Monthly Data Usage Trend",color="blue",fontsize=15,pad=10,fontstyle="italic")
plt.grid()
plt.show()
```
