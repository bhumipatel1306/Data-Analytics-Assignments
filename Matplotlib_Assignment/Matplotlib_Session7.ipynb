1.Use plt.subplot() to create a figure with 2 rows and 1 column of plots showing your daily step count and water intake (in litres) for the last 7 days. Plot both as line charts, one in each subplot.

```python
days = ['Mon', 'Tue', 'Wed', 'Thu','Fri', 'Sat', 'Sun']
steps = [4000, 5500, 5000, 7000,6500, 9000, 8000]
water = [1.5, 2.0, 1.8, 2.5,2.2, 3.0, 2.5]

plt.figure(figsize=(7, 7))

# First subplot - Daily Steps
plt.subplot(2, 1, 1)
plt.plot(days, steps,color='blue',marker='o',linewidth=2)
plt.title("Daily Step Count")
plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.grid(True, alpha=0.5)

# Second subplot - Water Intake
plt.subplot(2, 1, 2)
plt.plot(days, water,color='green',marker='o',linewidth=2)
plt.title("Daily Water Intake")
plt.xlabel("Days")
plt.ylabel("Water Intake (Litres)")
plt.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()
```

2.Create a 2x2 multi-panel figure using plt.subplots(). In each subplot, display the number of Instagram posts you made in each quarter (Q1, Q2, Q3, Q4) for the past year, using different chart types: line, bar, scatter, and pie.

```python
quarters = ['Q1', 'Q2', 'Q3', 'Q4']
posts = [12, 18, 15, 25]

fig, ax = plt.subplots(2, 2, figsize=(12, 8))

# 1. Line Chart
ax[0, 0].plot(quarters, posts,color='blue', marker='o',linewidth=2)
ax[0, 0].set_title("Instagram Posts - Line Chart")
ax[0, 0].set_xlabel("Quarters")
ax[0, 0].set_ylabel("Number of Posts")
ax[0, 0].grid(True, alpha=0.5)

# 2. Bar Chart
ax[0, 1].bar(quarters, posts,color='green', width=0.6)
ax[0, 1].set_title("Instagram Posts - Bar Chart")
ax[0, 1].set_xlabel("Quarters")
ax[0, 1].set_ylabel("Number of Posts")

# 3. Scatter Plot
ax[1, 0].scatter(quarters, posts,color='purple', s=100)
ax[1, 0].set_title("Instagram Posts - Scatter Plot")
ax[1, 0].set_xlabel("Quarters")
ax[1, 0].set_ylabel("Number of Posts")
ax[1, 0].grid(True, alpha=0.5)

# 4. Pie Chart
ax[1, 1].pie(posts,labels=quarters,autopct='%1.1f%%',startangle=90)
ax[1, 1].set_title("Instagram Posts - Pie Chart")
fig.suptitle("Instagram Quarterly Posts Analysis",fontsize=16)
plt.tight_layout()
plt.show()
```

3.Make a grid of 1 row and 3 columns using plt.subplots(), and plot the number of food orders you placed on Zomato, Swiggy, and Domino's over the last 6 months, one app per subplot. Share the y-axis among all subplots so you can easily compare order counts across apps.<br><br><em><strong>Hint:</strong> Use the 'sharey' parameter in plt.subplots().</em>

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
zomato = [5, 8, 6, 10, 7, 12]
swiggy = [4, 6, 8, 7, 9, 11]
dominos = [2, 3, 4, 5, 3, 6]
fig, ax = plt.subplots(
    1, 3,
    figsize=(15, 5),
    sharey=True
)

# First subplot - Zomato
ax[0].plot(months, zomato,color='red',marker='o',linewidth=2)
ax[0].set_title("Zomato Orders")
ax[0].set_xlabel("Months")
ax[0].set_ylabel("Number of Orders")
ax[0].grid(True, alpha=0.5)

# Second subplot - Swiggy
ax[1].plot(months, swiggy,color='orange',marker='o',linewidth=2)
ax[1].set_title("Swiggy Orders")
ax[1].set_xlabel("Months")
ax[1].grid(True, alpha=0.5)

# Third subplot - Domino's
ax[2].plot(months, dominos,color='blue',marker='o',linewidth=2)
ax[2].set_title("Domino's Orders")
ax[2].set_xlabel("Months")
ax[2].grid(True, alpha=0.5)

fig.suptitle("Food Delivery Apps - Monthly Orders")
plt.tight_layout()
plt.show()
```

4.Arrange a 2x2 grid of subplots using plt.subplots() to visualize your monthly mobile data usage (in GB) for four different apps: YouTube, Instagram, Spotify, and WhatsApp. Ensure all subplots share the same x-axis (months) for easy comparison.<br><br><em><strong>Hint:</strong> Use the 'sharex' parameter in plt.subplots().</em>

```python
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
youtube = [5, 8, 7, 10, 12, 15]
instagram = [3, 5, 6, 8, 9, 11]
spotify = [2, 3, 2, 4, 5, 6]
whatsapp = [1, 2, 2, 3, 3, 4]

fig, ax = plt.subplots(2, 2, figsize=(12, 8), sharex=True)

# YouTube
ax[0, 0].plot(months, youtube, marker='o', color='red')
ax[0, 0].set_title('YouTube Data Usage')
ax[0, 0].set_ylabel('Data Usage (GB)')
ax[0, 0].grid(True)

# Instagram
ax[0, 1].plot(months, instagram, marker='o', color='purple')
ax[0, 1].set_title('Instagram Data Usage')
ax[0, 1].set_ylabel('Data Usage (GB)')
ax[0, 1].grid(True)

# Spotify
ax[1, 0].plot(months, spotify, marker='o', color='green')
ax[1, 0].set_title('Spotify Data Usage')
ax[1, 0].set_xlabel('Months')
ax[1, 0].set_ylabel('Data Usage (GB)')
ax[1, 0].grid(True)

# WhatsApp
ax[1, 1].plot(months, whatsapp, marker='o', color='blue')
ax[1, 1].set_title('WhatsApp Data Usage')
ax[1, 1].set_xlabel('Months')
ax[1, 1].set_ylabel('Data Usage (GB)')
ax[1, 1].grid(True)

fig.suptitle('Monthly Mobile Data Usage by Apps')
plt.tight_layout()
plt.show()
```
