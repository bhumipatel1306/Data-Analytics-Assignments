2.Create a scatter plot showing 'amount spent on Zomato orders' vs 'number of orders placed' for 12 friends, and label each point with the friend's name.<br><br><em><strong>Hint:</strong> Use the 'annotate' function in matplotlib to add labels.</em>

```python
import matplotlib.pyplot as plt

friends = ['Amit', 'Rahul', 'Priya', 'Neha','Karan', 'Riya', 'Jay', 'Meet','Pooja', 'Ankit', 'Kavya', 'Rohit']
amount = [500, 1200, 800, 1500, 2000, 700,1800, 1000, 2500, 1300, 900, 2200]
orders = [2, 5, 3, 6, 8, 2,7, 4, 9, 5, 3, 8]

plt.figure(figsize=(10, 6))
plt.scatter(amount, orders, color='blue', s=80)

# Add friend's name to each point
for i in range(len(friends)):
    plt.annotate(friends[i],(amount[i], orders[i]),xytext=(5, 5),textcoords='offset points')

plt.title("Zomato Spending vs Number of Orders")
plt.xlabel("Amount Spent (₹)")
plt.ylabel("Number of Orders")
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

3.Add a trend line to your Zomato scatter plot by calculating the best-fit line using numpy's polyfit(), then plot it on top of the scatter plot.<br><br><em><strong>Hint:</strong> Use np.polyfit() to get the slope and intercept, then plot the line with plt.plot().</em>

```python
import matplotlib.pyplot as plt
import numpy as np

friends = ['Amit', 'Rahul', 'Priya', 'Neha','Karan', 'Riya', 'Jay', 'Meet','Pooja', 'Ankit', 'Kavya', 'Rohit']
amount = [500, 1200, 800, 1500, 2000, 700,1800, 1000, 2500, 1300, 900, 2200]
orders = [2, 5, 3, 6, 8, 2,7, 4, 9, 5, 3, 8]
plt.figure(figsize=(10, 6))
plt.scatter(amount, orders, color='blue', s=80)

# Add friend names
for i in range(len(friends)):
    plt.annotate(
        friends[i],
        (amount[i], orders[i]),
        xytext=(5, 5),
        textcoords='offset points'
    )

# Calculate best-fit line
m, c = np.polyfit(amount, orders, 1)

# Create trend line
x_line = np.array([min(amount), max(amount)])
y_line = m * x_line + c

# Plot trend line
plt.plot(x_line, y_line, color='red',linestyle='--', linewidth=2,label='Trend Line')

plt.title("Zomato Spending vs Number of Orders")
plt.xlabel("Amount Spent (₹)")
plt.ylabel("Number of Orders")
plt.legend()
plt.grid(True, alpha=0.5)
plt.show()
```

4.Given a dataset of 'YouTube watch hours' and 'Spotify listening hours' for 15 users, use numpy polyfit to fit a regression line, plot both the scatter and the trend line, and print the equation of the line on the plot.

```python
import matplotlib.pyplot as plt
import numpy as np

# Dataset of 15 users
youtube = [10, 15, 20, 25, 30,35, 40, 45, 50, 55,60, 65, 70, 75, 80]
spotify = [5, 8, 12, 15, 18,20, 25, 28, 30, 35,38, 42, 45, 48, 52]

# Scatter Plot
plt.figure(figsize=(10, 6))
plt.scatter(youtube, spotify,color='blue', s=80,label='Users')

# Calculate regression line
m, c = np.polyfit(youtube, spotify, 1)

# Calculate Y values
x_line = np.array([min(youtube), max(youtube)])
y_line = m * x_line + c

# Plot regression line
plt.plot(x_line, y_line,
         color='red',
         linestyle='--',
         linewidth=2,
         label='Regression Line')

# Print equation on graph
equation = f"y = {m:.2f}x + {c:.2f}"

plt.text(12, 48, equation,
         fontsize=12,
         color='red')

plt.title("YouTube Watch Hours vs Spotify Listening Hours")
plt.xlabel("YouTube Watch Hours")
plt.ylabel("Spotify Listening Hours")

plt.legend()
plt.grid(True, alpha=0.5)

plt.show()

# Print equation in output
print("Regression Equation:", equation)
```
