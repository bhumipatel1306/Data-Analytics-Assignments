```python
import matplotlib.pyplot as plt
```

1.Use plt.bar() to create a vertical bar chart showing the number of daily orders for 5 popular food delivery apps (e.g., Zomato, Swiggy, Uber Eats, Dunzo, Domino's) and set a custom color for each bar.

```python
delivery_app = ["Zomato","Swiggy","Uber Eats","Dunzo","Domino's"]
orders = [25,15,13,22,18]

plt.bar(delivery_app,orders,color=)
plt.title("Number of Daily Orders",pad=10,fontstyle="italic",fontweight="bold",color="g")
plt.xlabel("Delivery App",labelpad=10,fontsize=10,color="black")
plt.ylabel("Number of Orders",labelpad=10,fontsize=10,color="black")
plt.show()
```

2.Create a horizontal bar chart using plt.barh() to display the number of followers for 6 different Instagram influencers; add a title and label the x-axis as 'Followers (in millions)'.

```python
influencers = ["Influencer A", "Influencer B","Influencer C", "Influencer D","Influencer E", "Influencer F"]
followers = [2.5, 5.0, 3.8, 7.2, 4.5, 6.0]

plt.barh(influencers,followers)
plt.title("Number of Followers",pad=10,fontstyle="italic",fontweight="bold",color="g")
plt.xlabel("Followers (in millions)",labelpad=10,fontsize=10,color="r")
plt.ylabel("Influencers",labelpad=10,fontsize=10,color="r")
plt.show()
```

3.Modify your vertical bar chart to display the exact order count as a text label above each bar using plt.text().<br><br><em><strong>Hint:</strong> Use a for loop to iterate through the bars and place the text labels at the correct height.</em>

```python
months = ["January", "February", "March", "April", "May"]
orders = [25,15,13,22,18]

plt.figure(figsize=(9, 5))
plt.bar(months, orders, color="#DA4494")
plt.title("Monthly Zomato Orders")
plt.xlabel("Months")
plt.ylabel("Number of Orders")

for i in range(len(months)):
    plt.text(i, orders[i] + 0.2,str(orders[i]),ha="center")
plt.show()
```

4.Change the width of the bars in your food delivery app chart to make them thinner, and adjust the spacing so the bars do not overlap.<br><br><em><strong>Constraint:</strong> Use the 'width' parameter in plt.bar() and experiment with values less than 0.8.</em>

```python
delivery_app = ["Zomato","Swiggy","Uber Eats","Dunzo","Domino's"]
orders = [25,15,13,22,18]

plt.bar(delivery_app,orders,width=0.6)
plt.title("Number of Daily Orders",pad=10,fontstyle="italic",fontweight="bold",color="g")
plt.xlabel("Delivery App",labelpad=10,fontsize=10,color="black")
plt.ylabel("Number of Orders",labelpad=10,fontsize=10,color="black")
plt.show()
```
