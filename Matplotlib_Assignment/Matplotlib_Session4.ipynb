1.Use plt.pie() to visualize the percentage of time you spend on 5 different apps in a day (e.g., Instagram, YouTube, WhatsApp, Zomato, Spotify) and display the percentage values on each segment using the autopct parameter.

```python
from enum import auto
apps = ['Instagram', 'YouTube', 'WhatsApp','Zomato', 'Spotify']
hours = [4, 5, 3, 2, 1]

plt.pie(hours,labels=apps,autopct="%1.1f%%")
plt.title("Daily App Usage Percentage",color="r")
plt.show()
```

2.Create a pie chart showing the distribution of your monthly online spending across 4 categories: Food Delivery, Shopping, Entertainment, and UPI Payments, and use the explode parameter to highlight the category with the highest spending.<br><br><em><strong>Hint:</strong> Set a higher explode value for the segment you want to pull out from the pie.</em>

```python
categories = ['Food Delivery','Shopping','Entertainment','UPI Payments']
spending = [3000, 5000, 2000, 4000]
colors=["#1D3557","#457B9D","#A8DADC","#E63946"]

plt.pie(spending,labels=categories,colors=colors,startangle=20,explode=(0,0.08,0,0),
        wedgeprops={'edgecolor':'white','linewidth':1})
plt.title("Monthly Online Spending",color="blue",fontstyle="oblique",fontsize=15)
plt.show()
```

3.Transform your pie chart into a donut chart to show the same spending data by adding a white circle at the center using plt.Circle and ax.add_artist.

```python
categories = ['Food Delivery','Shopping','Entertainment','UPI Payments']
spending = [3000, 5000, 2000, 4000]
colors = ['#264653', '#287271', '#2A9D8F', '#E9C46A']


fig, ax = plt.subplots(figsize=(7, 7))
ax.pie(
    spending,
    labels=categories,
    colors=colors,
    explode= (0, 0.15, 0, 0),
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}
)

# Create White Circle
centre_circle = plt.Circle(
    (0, 0),
    0.50,
    color='white'
)

ax.add_artist(centre_circle)
plt.title('Monthly Online Spending Donut Chart',fontsize=14,fontweight='bold',pad=20)
ax.set_aspect('equal')
plt.show()
```

4.Suppose you want to compare the market share of 4 IPL cricket teams based on the number of Instagram followers. Build a donut chart where each segment shows the team name and follower percentage, and the chart title clearly states 'IPL Teams Instagram Followers Share'.<br><br><em><strong>Constraint:</strong> Use both labels and autopct to display information on the chart.</em>

```python
teams = ['RCB', 'CSK', 'MI', 'KKR']
followers = [24.8, 23, 20, 8.1]
colors = ['#A71930', '#F9CD05', '#004BA0', '#3A225D']
plt.figure(figsize=(6, 6))

# Donut Chart
plt.pie(
    followers,
    labels=teams,
    colors=colors,
    autopct='%1.2f%%',
    startangle=90,
    wedgeprops={
        'width': 0.5,
        'edgecolor': 'white',
        'linewidth': 2
    }
)
plt.text(0, 0,'75.9M\nTotal Followers',ha='center',va='center',fontsize=13,fontweight='bold')
plt.title('IPL Teams Instagram Followers Share',fontsize=15,fontweight='bold',pad=20)
plt.show()
```
