```python
import matplotlib.pyplot as plt
```

1.Create a simple bar chart showing the number of orders for 5 different Zomato restaurants and save the chart as a PNG file using plt.savefig() with a DPI of 150.

```python
restaurants = ['Dominos', 'McDonalds', 'La Pinoz','Burger King', 'Subway']
orders = [450, 600, 350, 500, 400]

plt.figure(figsize=(9, 5))
plt.bar(restaurants, orders, width=0.6)
plt.title('Monthly Orders - Zomato Restaurants')
plt.xlabel('Restaurant Names')
plt.ylabel('Number of Orders')
plt.grid(axis='y', linestyle='--', alpha=0.5)
# Save chart as PNG
plt.savefig('zomato_orders.png', dpi=150, bbox_inches='tight')
plt.show()
```

2.Generate a line plot that displays your daily step count for one week (7 days) and export it as an SVG file using plt.savefig().<br><br><em><strong>Hint:</strong> Change the file extension in the filename argument to save as SVG.</em>

```python
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday','Friday', 'Saturday', 'Sunday']
steps = [4000, 5500, 6500, 5000, 7500, 9000, 8000]

plt.figure(figsize=(9, 5))
plt.plot(days, steps, marker='o', color='green',linewidth=2, markersize=7)
plt.title('Daily Step Count - One Week')
plt.xlabel('Days')
plt.ylabel('Number of Steps')
plt.grid(True)
plt.savefig('weekly_steps.svg', format='svg')
plt.show()
```

3.Build a pie chart showing the percentage of time you spend on 4 different apps (Instagram, YouTube, WhatsApp, Spotify) in a day and save the chart as a PDF using plt.savefig().

```python
apps = ["Instagram", "YouTube", "WhatsApp", "Spotify"]
hours = [3, 2, 1, 2]

plt.figure(figsize=(6, 6))
plt.pie(hours,labels=apps,autopct="%1.1f%%",startangle=90,explode=(0.1, 0, 0, 0))
plt.title("Daily App Usage Percentage")
plt.savefig("app_usage.pdf", format="pdf", dpi=150, bbox_inches="tight")
plt.show()
```
