1.Create a line chart using Matplotlib to show the number of daily active users for a music streaming app (like Spotify) over 12 months, and use plt.annotate() to label the month with the highest user count.

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun','Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
users = [120, 135, 150, 145, 170, 180,165, 190, 210, 200, 230, 250]

# Find highest users
max_users = max(users)
max_index = users.index(max_users)

plt.figure(figsize=(8, 5))
plt.plot(months, users, marker='o', color='green',linewidth=2, label='Daily Active Users')

# Annotate highest user count
plt.annotate(
    f'Highest: {max_users}K',
    xy=(max_index, max_users), #Target point (11,250)
    xytext=(max_index-2, max_users-30), # Text ponit (9,220)
    arrowprops=dict(arrowstyle='->', color='red'),fontsize=12,color='red')
plt.title('Spotify Daily Active Users Over 12 Months')
plt.xlabel('Months')
plt.ylabel('Daily Active Users (in thousands)')
plt.grid(True)
plt.legend()
plt.show()
```

2.Plot a bar chart showing the number of orders per month for a food delivery app (like Zomato), and add an arrow annotation using plt.annotate() to highlight the month with the lowest orders.<br><br><em><strong>Hint:</strong> Use the 'arrowprops' parameter to draw the arrow pointing to the bar.</em>

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun','Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
orders = [450, 500, 600, 550, 700, 650,400, 750, 800, 720, 680, 900]

# Find lowest orders
min_orders = min(orders)
min_index = orders.index(min_orders)

plt.figure(figsize=(8, 5))
plt.bar(months, orders, width=0.6)

# Annotate lowest orders
plt.annotate(
    f'Lowest: {min_orders}',
    xy=(min_index, min_orders),
    xytext=(min_index+1, min_orders+150),
    arrowprops=dict(arrowstyle='->', color='red'),fontsize=12,color='black')
plt.title('Monthly Food Delivery Orders')
plt.xlabel('Months')
plt.ylabel('Number of Orders')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
```

3.Create a pie chart representing the percentage of different payment methods (UPI, Card, Wallet, Cash) used on an e-commerce app (like Flipkart), and add a legend positioned at the lower right corner of the plot.

```python
payment_methods = ['UPI', 'Card', 'Wallet', 'Cash']
payments = [50, 25, 15, 10]
#colors = ['skyblue', 'orange', 'lightgreen', 'pink']
#colors = ["#1F4E78","#2E75B6","#70AD47","#A5A5A5"]
colors = ["#4B0082", "#6A0DAD", "#9370DB", "#D8BFD8"]

plt.figure(figsize=(6, 6))
plt.pie(
    payments,
    labels=payment_methods,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90
)

plt.legend(loc='lower right', title='Payment Methods')
plt.title('E-commerce Payment Methods Distribution')
plt.show()
```

4.Given a scatter plot of movie ratings vs. number of reviews for movies on a streaming platform (like BookMyShow), use plt.annotate() to add a text box highlighting the movie with the highest rating.<br><br><em><strong>Hint:</strong> Set the 'bbox' parameter in plt.annotate() to create a visible text box.</em>

```python
movies = ['Ramayana', 'Krishnavataram', 'Hanu Man','777 Charlie', 'Kantara', 'Dear Comrade']
ratings = [9.5, 9.2, 8.8, 9.0, 8.7, 8.4]
reviews = [15000, 12000, 9000, 11000, 13000, 7000]

# Find highest rating
max_rating = max(ratings)
max_index = ratings.index(max_rating)

plt.figure(figsize=(8, 6))
plt.scatter(reviews, ratings, color='#2E75B6',s=120, alpha=0.8)

plt.annotate(
    f'{movies[max_index]} ({max_rating})',
    xy=(reviews[max_index], max_rating),
    xytext=(reviews[max_index]-2000, max_rating-0.5),
    bbox=dict(boxstyle='round', facecolor='lightyellow',
              edgecolor='#1F4E78'),
    arrowprops=dict(arrowstyle='->', color='red'),
    fontsize=11
)
plt.title('Latest Indian Movies - Ratings vs Reviews')
plt.xlabel('Number of Reviews')
plt.ylabel('Movie Ratings')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
