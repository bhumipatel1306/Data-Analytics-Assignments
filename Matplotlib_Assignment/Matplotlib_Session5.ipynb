1.Use plt.hist() to plot a histogram of delivery times (in minutes) for 30 Swiggy orders: [32, 28, 45, 30, 36, 40, 29, 34, 38, 31, 27, 41, 33, 35, 39, 37, 44, 30, 29, 32, 36, 28, 43, 31, 35, 40, 38, 33, 30, 42]. Show the plot window.

```python
delivery_times = [
    32, 28, 45, 30, 36, 40, 29, 34, 38, 31,
    27, 41, 33, 35, 39, 37, 44, 30, 29, 32,
    36, 28, 43, 31, 35, 40, 38, 33, 30, 42]
plt.figure(figsize=(7, 5))

plt.hist(delivery_times,bins=5,color='#287271')
plt.xlabel('Delivery Time (Minutes)')
plt.ylabel('Number of Orders')
plt.title('Swiggy Order Delivery Time Distribution',fontsize=14,fontweight='bold')
plt.show()
```

2.Plot a histogram of IPL match scores (runs per team in a match) using plt.hist() and experiment with bins=5, bins=10, and bins=15. Observe and describe how changing the number of bins affects the shape of the histogram.<br><br><em><strong>Hint:</strong> Use any sample array of 50-60 scores between 120 and 220.</em>

```python
scores = [
    145, 167, 189, 175, 198, 210, 156, 182, 195, 220,
    134, 160, 178, 188, 205, 215, 149, 170, 185, 199,
    125, 155, 165, 180, 192, 207, 218, 140, 172, 190,
    200, 212, 130, 150, 168, 183, 196, 208, 216, 142,
    158, 177, 186, 202, 214, 128, 162, 173, 193, 204,
    219, 147, 166, 179, 187
]
plt.figure(figsize=(15, 5))

# Histogram 1: bins=5
plt.subplot(1, 3, 1)
plt.hist(scores,bins=5,color='#264653',edgecolor='white')
plt.title('IPL Scores (bins=5)')
plt.xlabel('Runs')
plt.ylabel('Frequency')


# Histogram 2: bins=10
plt.subplot(1, 3, 2)
plt.hist(scores,bins=10,color='#287271',edgecolor='white')
plt.title('IPL Scores (bins=10)')
plt.xlabel('Runs')
plt.ylabel('Frequency')


# Histogram 3: bins=15
plt.subplot(1, 3,3)
plt.hist(scores,bins=15,color='#E9C46A',edgecolor='black')
plt.title('IPL Scores (bins=15)')
plt.xlabel('Runs')
plt.ylabel('Frequency')
plt.suptitle('IPL Match Scores Distribution',fontsize=16,fontweight='bold')
plt.tight_layout()
plt.show()
```

3.Given an array of daily step counts from a fitness tracker app for one month, plot two histograms using plt.hist(): one showing frequency (default), and one showing density=True. Compare the y-axis labels and explain the difference.

```python
steps = [
    4500, 5200, 6800, 7200, 8100,
    5600, 6200, 7500, 8900, 9500,
    4800, 5900, 6400, 7800, 8500,
    9200, 10500, 11200, 6800, 7300,
    8200, 9700, 5500, 6100, 7600,
    8800, 9900, 11500, 12500, 7000
]

# Figure Size
plt.figure(figsize=(13, 5))

# First Histogram - show Frequency
plt.subplot(1, 2, 1)
plt.hist(steps,bins=5,color='#264653',edgecolor='white')
plt.title('Daily Steps - Frequency')
plt.xlabel('Daily Step Counts')
plt.ylabel('Frequency (Number of Days)')

# Second Histogram - Density  show normalized probability distribution
plt.subplot(1, 2, 2)
plt.hist(steps,bins=5,density=True,color='#2A9D8F',edgecolor='white')
plt.title('Daily Steps - Density')
plt.xlabel('Daily Step Counts')
plt.ylabel('Probability Density')

plt.suptitle('Fitness Tracker Daily Steps Distribution',fontsize=15,fontweight='bold')
plt.tight_layout()
plt.show()

print("The frequency histogram shows the number of days in each step-count interval, while the density histogram shows the normalized probability distribution. In a density histogram, the total area of all bars equals 1.")
```

4.Simulate the distribution of song durations (in seconds) from your Spotify playlist by generating 50 random numbers between 120 and 300. Plot a histogram using plt.hist() and set the color to 'purple' and add the title 'Spotify Song Duration Distribution'.
