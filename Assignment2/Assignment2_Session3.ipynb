3.Using a DataFrame of food orders from Zomato with columns 'restaurant', 'order_amount', and 'delivery_time', filter and display only those orders where 'order_amount' is greater than 500 using a boolean condition.

```python
import pandas as pd

data = {
    "restaurant": ["Dominos", "McDonalds", "La Pinoz", "Subway", "Burger King","Pizza Hut", "Starbucks", "Biryani House", "KFC", "Honest"],
    "order_amount": [350, 650, 800, 450, 720,550, 900, 1200, 480, 600],
    "delivery_time": [30, 40, 45, 25, 35,50, 30, 55, 40, 35]
}

dff = pd.DataFrame(data)
#print(dff)
filter = dff[dff['order_amount'] > 500]
print(filter)
```

4.Given a DataFrame of Spotify songs with columns 'song', 'artist', 'streams', and 'duration', use the query() method to select all songs with more than 1,000,000 streams and duration less than 180 seconds.<br><br><em><strong>Hint:</strong> Use the query syntax: query('streams > 1000000 and duration < 180')</em>

```python
import pandas as pd

Playlist= {
    "song": ["Darkhaast","Khat","Ganga ke Kinare","Vaaroon","Tum Hi Ho","Apna Bana Le","O Maahi"],
    "artist": ["Arijit Singh","Navjot Ahuja","Sagar","Anand Bhaskar","Arijit Singh","Arijit Singh","Arijit Singh"],
    "streams": [1500000,1200000,900000,1800000,1400000,1300000,1600000],
    "duration": [170,165,190,160,175,179,170]
}
df = pd.DataFrame(Playlist)
#print(df)
filter = df.query('streams > 1000000 and duration < 180')
print(filter)
```

5.Suppose you have a DataFrame of IPL cricket matches with columns 'team1', 'team2', 'winner', and 'total_runs'. Write code to filter and display all matches where 'total_runs' is between 180 and 220 using boolean indexing.<br><br><em><strong>Constraint:</strong> Do not use the query() method for this task.</em>

```python
import pandas as pd

cricket = {
    "team1": ["CSK", "MI", "RCB", "KKR", "GT", "DC"],
    "team2": ["MI", "RCB", "KKR", "GT", "DC", "CSK"],
    "winner": ["CSK", "MI", "RCB", "KKR", "GT", "DC"],
    "total_runs": [195, 210, 175, 220, 205, 240]
}

df = pd.DataFrame(cricket)
#print(df)
filter = df[(df["total_runs"] >= 180) & (df["total_runs"] <= 220)]
print(filter)
```
