1.Install Matplotlib in your Python environment and write a script named hello_plot.py that creates a simple line plot showing the number of daily steps you walked over the past 7 days.

```python
from matplotlib.lines import lineStyles
days = ["Monday", "Tuesday", "Wednesday","Thursday", "Friday", "Saturday", "Sunday"]
steps = [3000, 4500, 5000, 4000, 6500, 8000, 7000]

plt.plot(days,steps,marker="s")
plt.grid(True,linestyle="--",alpha=0.7)
plt.xlabel("days",color="r",fontsize=13)
plt.ylabel("steps",color="r",fontsize=13)
plt.title("My steps over the week")
plt.show()
```

2.Create a script called insta_likes_plot.py that uses Matplotlib to plot the number of likes you received on your last 10 Instagram posts (use made-up data if needed), labeling the axes appropriately.

```python
posts = ["Post 1", "Post 2", "Post 3", "Post 4", "Post 5","Post 6", "Post 7", "Post 8", "Post 9", "Post 10"]
likes = [120, 250, 180, 400, 350, 500, 450, 600, 750, 900]

plt.plot(posts, likes,color=(1,0,0,0.8),marker='o',markersize=10,markerfacecolor='yellow',markeredgecolor='black',alpha=0.6,label="Like",linestyle='--',linewidth=3)
plt.xlabel("Instagram Posts",color="b",fontsize=10)
plt.ylabel("Number of Likes",color="b",fontsize=10)
plt.title("Instagram Post Likes")
plt.show()
```

3.Write a Python script that creates a Matplotlib figure with two Axes: one showing your daily Spotify listening hours for a week and the other showing your daily YouTube viewing hours for the same week. Display both plots in a single figure.

```python
df = pd.DataFrame({
    "Day": ["Monday", "Tuesday", "Wednesday", "Thursday","Friday", "Saturday", "Sunday"],
    "Spotify_Hours": [2, 1.5, 3, 2.5, 4, 5, 3.5],
    "YouTube_Hours": [1, 2, 1.5, 3, 2.5, 4, 3]
})

fig, ax = plt.subplots(
    1,2,
    figsize=(20,8),
    facecolor="lightblue",
    constrained_layout=True
)

ax[0].plot(df["Day"], df["Spotify_Hours"], marker="o")

ax[1].plot(df["Day"], df["YouTube_Hours"], marker="o")
plt.show()
```

4.Add axis labels, a title, and a grid to your plot from the previous task to clearly communicate the information, mimicking the style of analytics dashboards in apps like Instagram Insights.<br><br><em><strong>Hint:</strong> Use set_xlabel(), set_ylabel(), set_title(), and grid() methods on the Axes object.</em>

```python
posts = ["Post 1", "Post 2", "Post 3", "Post 4", "Post 5","Post 6", "Post 7", "Post 8", "Post 9", "Post 10"]
likes = [120, 250, 180, 400, 350, 500, 450, 600, 750, 900]

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(posts, likes, marker='o')

# Axis Labels
ax.set_xlabel("Instagram Posts")
ax.set_ylabel("Number of Likes")

# Title
ax.set_title("Instagram Post Likes Insights")

# Grid
ax.grid(True)
plt.show()
```
